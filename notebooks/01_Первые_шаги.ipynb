### Как устроен ноутбук

Ноутбук - это последовательность **ячеек**: текстовых (как эта) и кодовых.
Кодовая ячейка выполняется по **Shift + Enter**, результат появляется под ней.

- За ноутбуком стоит **ядро** (kernel) — запущенный процесс Python, который помнит всё,
  что вы в нём создали. Переменная из ячейки 5 доступна в ячейке 20.
- Ядро помнит порядок **выполнения**, а не порядок ячеек на экране. Если пропустить ячейку
  или выполнить их вразнобой, состояние перестанет соответствовать тому, что вы видите.
  Это причина большей части «загадочных» сбоев в ноутбуках.
- Если состояние запуталось: **Runtime → Restart runtime** (в Colab) и выполнить ячейки
  сверху вниз.

## 1. Python за пять минут

**Python** — высокоуровневый интерпретируемый язык общего назначения с динамической
типизацией. Создан Гвидо ван Россумом в 1991 году, актуальная ветка — Python 3.

| Свойство | Что это значит на практике |
|---|---|
| Интерпретируемый | Нет отдельного этапа компиляции: интерпретатор читает и исполняет код по мере продвижения. Быстрый цикл «написал → запустил», но многие ошибки обнаруживаются только в момент, когда исполнение дошло до строки |
| Динамическая типизация | Тип принадлежит **значению**, а не переменной. Объявлять типы не нужно, плата за это — часть ошибок ловится только при запуске |
| Высокоуровневый | Управление памятью, указатели и прочее скрыты от вас |
| Значимые отступы | Структура блоков задаётся отступами, а не скобками |
| Богатая экосистема | Обширная стандартная библиотека плюс сотни тысяч пакетов в репозитории PyPI |

**Чем платим:** скорость исполнения на 1–2 порядка ниже, чем у C или Rust. Для наших задач
это несущественно: тяжёлые вычисления в Python принято отдавать библиотекам, у которых
внутри, например, C.

## 2. Инженерное мышление

А. Спрол «Думай как программист», главы 1, 8

### 2.1. Восемь подходов к решению задач

**1. Планируйте решение.** Всегда должен быть план. На старте у вас может не быть 
ни малейшего представления, как решать задачу. Но у вас должен быть план того, 
**как вы будете искать решение**.

Эйзенхауэр: «Я всегда убеждался, что планы бесполезны,
но без планирования не обойтись». План почти наверняка изменится по ходу решения,
но при планировании вы разбираетесь в задаче.

У плана есть и второе назначение — **промежуточные цели**. Без плана цель одна,
«решить задачу», и до самого конца вы не чувствуете прогресса. 

**2. Переформулируйте задачу.** Классическая головоломка: крестьянину нужно перевезти
через реку лисицу, гуся и мешок кукурузы. В лодке помещается он сам и один объект.
Лисицу нельзя оставить с гусем, гуся — с кукурузой.

Большинство людей не решают её, потому что не рассматривают возможность отвезти
объект **обратно**. В условии это не запрещено — просто не сказано.
Отсюда приём: прежде чем решать, выпишите формально:

- **Ограничения** — что нельзя, что обязано выполняться.
- **Операции** — что вы вообще можете сделать.

Причём операции стоит формулировать максимально обобщённо. Список «перевезти лисицу
на тот берег / перевезти гуся / перевезти кукурузу» решения не подсказывает. Список
«переплыть с берега на берег / загрузить объект / выгрузить объект» — подсказывает,
потому что в нём нет неявного допущения о направлении. Спрол формулирует это так:
**если вы не знаете всех доступных вам операций, вы, возможно, не сможете решить задачу.**

Работает и более простой вариант приёма: перескажите условие своими словами вслух
или письменно. Программисты часто идут обсуждать задачу к коллеге не потому, что
у коллеги есть решение, а потому, что проговаривание само наводит на мысли.
Переформулирование — то же обсуждение, в котором вы играете обе роли.

> Общий вывод из этого примера: **обдумывание задачи бывает продуктивнее обдумывания
> решения.** Во многих случаях правильная постановка задачи и есть решение.



**3. Делите задачу.** Разбиение на подзадачи уменьшает объём исходной задачи. Аналогия из книги: разложить по алфавиту 100 карточек тяжелее, чем четыре раза разложить по 25, потому что при вставке каждой следующей
карточки приходится просматривать все уже разложенные. Работа растёт быстрее, чем
количество элементов («квадратичная сложность»).

С кодом то же самое: `if` внутри `while` внутри `for` понять и написать тяжелее,
чем те же три конструкции по отдельности. Признак хорошей декомпозиции — любую
часть можно написать и проверить, не держа в голове остальные.

**4. Начните с того, что знаете.** Разделив задачу, сделайте сначала те части, которые
умеете. Работающая часть решения подсказывает идеи для остального. Аналогия Спрола: прежде чем ехать в строительный магазин, попробуйте
сделать ремонт теми инструментами, что уже есть дома.

**5. Упрощайте задачу.** Если задача не поддаётся, снимите с неё часть требований
и решите то, что осталось.

Пример из книги: даны точки в трёхмерном пространстве, нужно найти две ближайшие.
Не получается — упростите. Можно убрать измерения: пусть точки лежат на прямой,
тогда задача сводится к поиску двух чисел с наименьшей разностью. А можно оставить
три измерения, но взять ровно три точки — тогда останется сравнить A с B, B с C и A с C.
Обратите внимание: **два упрощения снимают разные трудности**, и, решив обе упрощённые
версии, вы, как правило, уже понимаете, как собрать полную.

Упрощение точно локализует место, где вы застряли. Вместо «вот моя программа, она
не работает, почему?» получается «я умею считать расстояние между точками и умею
сравнивать расстояния, но не знаю, как перебрать все пары».

**6. Ищите аналогии.** Главный вопрос при встрече с новой задачей: **«какую похожую
задачу я уже решал?»** Это самый сильный ускоритель и одновременно самый трудный
для развития навык, потому что он требует архива собственных решённых задач.

Отсюда предупреждение Спрола: находить чужой похожий код и подгонять под себя —
плохая замена аналогии. Во-первых, правильно изменить программу, которую вы не
понимаете до конца, почти невозможно. Во-вторых, каждая написанная **вами** программа —
это будущий источник аналогий. Чем больше вы опираетесь на чужой код сейчас, тем
больше придётся опираться на него потом.

**7. Экспериментируйте.** Эксперимент — это не угадывание. Угадывание: написать
что-нибудь и посмотреть, не заработает ли. Эксперимент: сформулировать **гипотезу**
(«если сделать так, получится вот это»), проверить и сравнить результат с ожиданием.
Полезны оба исхода, но особенно — несовпадение: оно показывает, что ваша модель
языка в этом месте неверна.

Ноутбук — идеальный инструмент для этого: отдельная ячейка стоит секунду. Не уверены,
что делает метод или как поведёт себя операция, — проверьте на трёх строках.

**8. Не расстраивайтесь.** Раздражённый человек думает хуже, работает дольше. Наблюдение Спрола: программист злится не на код —
злиться на текст на экране бессмысленно, — а на себя. И, позволяя себе злиться,
он фактически выдаёт себе оправдание будущих неудач: «я бы решил, если бы не нервничал».

Заметили, что третий раз меняете код без гипотезы, —
остановитесь, вернитесь к плану, упростите задачу или сделайте перерыв и вернитесь позже со свежей головой. 



> В судоку разумно начинать не с любой пустой клетки, а с той, где вариант всего один:
> например, в области 3×3 уже заполнено восемь клеток. Заполнив её, вы добавляете
> ограничений соседям — и появляются новые клетки с единственным вариантом.
>
> В искусственном интеллекте это называют эвристикой «наиболее ограниченной переменной».
> Ограничения делают задачу трудной, но они же сокращают перебор. Ищите то место
> задачи, где свободы меньше всего, — и начинайте оттуда.

### 2.2. Схема решения задачи

```
ВХОД  →  ОБРАБОТКА  →  ВЫХОД
```

Почти любая программа, которую вы напишете, устроена именно так.
Начинайте разбор задачи с того, что дано и что нужно найти: если вы
не можете назвать вход и выход, вы ещё не поняли задачу и писать код рано.


### Пример: от формулировки до плана

**Задача.** Магазин продаёт товар по фиксированной цене. Покупателю полагается скидка,
на итог начисляется налог. Нужно посчитать и показать сумму заказа.

Проходим по приёмам 2 и 3 — переформулируем и разделим.

- **Вход:** цена за единицу, количество, доля скидки, ставка налога.
- **Выход:** сумма без скидки, размер скидки, размер налога, итог к оплате.
- **Ограничения:** скидка считается от суммы без налога, налог — от суммы после скидки;
  деньги выводим с двумя знаками после запятой.
- **Операции, которые нам доступны:** умножить, вычесть, сложить, вывести на экран.
  Список короткий — и это хороший признак: задача целиком лежит внутри того,
  что мы уже умеем (приём 4).

Теперь план. Записываем его комментариями


````python
# INPUT: unit price, quantity, discount rate, tax rate

# PROCESS:
#   1. subtotal = price * quantity
#   2. discount = subtotal * discount rate
#   3. taxable  = subtotal - discount
#   4. tax      = taxable * tax rate
#   5. total    = taxable + tax

# OUTPUT: subtotal, discount, tax, total (two decimal places)
````

План занял шесть строк, и в нём нет ни одного вопроса «а как это сделать?».
К концу занятия мы заменим эти комментарии кодом.
Виден смысл промежуточных целей: готовых пунктов плана будет пять,
и каждый можно проверить отдельно, не дожидаясь конца работы.

### 2.3. Мастер-план

Восьмая глава книги посвящена тому, во что складываются приёмы из главы первой.
Спрол называет это **мастер-планом**: набор рабочих правил, составленный под ваши
собственные сильные и слабые стороны и применяемый к каждой задаче. Аналогия из
книги — тренер, который строит игру под конкретного игрока
с его навыками и травмами.

Первый шаг — найти свои слабые места:

- **на уровне кодирования** — повторяющиеся ошибки при написании кода. Классические
  примеры: ошибка «на единицу» (цикл выполняется на один раз больше или меньше нужного),
  путаница `=` и `==`, забытое преобразование типа;
- **на уровне проектирования** — проблемы на этапе обдумывания задачи. Список из книги:
  слишком усложнённые решения; трудно начать; не тестирую; чрезмерная уверенность
  в собственном коде; пробел в конкретной области (рекурсия, классы, указатели).

Дальше план строится **под** эти слабости. Не тестируете — сделайте тест обязательным
пунктом плана для каждого куска кода. Трудно начать — сознательно упростите задачу
и начните с черновой версии, запланировав её переписать. Усложняете —
заложите в план отдельный этап упрощения уже работающего решения.

Хороший план не требует борьбы с собственными особенностями. Спрол,
зная за собой нетерпение («попробую-ка запустить и посмотрю, что будет»), не заставляет
себя не торопиться, а держит две версии программы: черновик, где можно всё,
и чистовик, куда попадает только проверенный код. 


> ### Журнал ошибок
>
> Главная мысль главы 8: исправив ошибку, спросите себя не «что здесь
> сломалось», а **какой общий принцип я нарушил**.
>
> Пример из книги. Функция считает среднее положительных чисел массива и падает,
> если положительных чисел не оказалось: деление на ноль. Частный вывод — «проверять
> делитель на ноль» — почти бесполезен: делить на ноль вы будете редко. Общий вывод —
> «я не рассматриваю особые случаи входных данных» — работает всегда и заодно
> напоминает про пустые данные, про значения вне диапазона и про всё остальное.
>
> Заведите файл или страницу в тетради с двумя колонками: **что произошло** и
> **какой принцип нарушен**. Через месяц вы увидите свои три-четыре типовые ошибки.

## 3. Переменные и присваивание

**Переменная — это имя, связанное со значением.** Связь создаётся оператором `=`.


````python
unit_price = 250.0
quantity = 4

print(unit_price, quantity)
````

Знак `=` в программировании — не «равно» из математики, а команда: **«свяжи это имя
с этим значением»**. Отсюда конструкция, которая в математике была бы ложной:


````python
quantity = quantity + 1
print(quantity)
````

Читается справа налево: вычислить `quantity + 1` (получится 5), затем связать имя
`quantity` с результатом. Прежняя связь теряется.

Такое встречается настолько часто, что для него есть сокращённая запись:


````python
quantity += 1     # то же самое, что quantity = quantity + 1
quantity -= 2
quantity *= 3
print(quantity)
````

**Проверьте себя.** Что напечатает следующая ячейка? Сначала ответьте, потом запускайте.


````python
a = 5
b = a
a = 10

print(a, b)
````

Печатается `10 5`. Строка `b = a` связала имя `b` со **значением**, которое имя `a`
обозначало в тот момент. Это не уравнение и не постоянная связь между именами:
последующее переприсваивание `a` про `b` ничего не знает.

Если хотите увидеть это своими глазами — вставьте код в визуализатор
[pythontutor.com](https://pythontutor.com/): он рисует имена и значения по шагам.

## 4. Типы данных

У каждого значения есть тип. Тип определяет, какие операции с этим значением осмысленны.
Базовых типов, которых хватит на первые занятия, пять:


````python
print(type(42))        # целое число
print(type(3.14))      # число с плавающей точкой
print(type("42"))      # строка
print(type(True))      # логическое значение
print(type(None))      # отсутствие значения
````

Обратите внимание: `42` и `"42"` — разные вещи. Первое — число, с ним можно считать.
Второе — текст, состоящий из двух символов.


````python
value = "42"

print(value + value)               # строки склеиваются
print(int(value) + int(value))     # числа складываются
````

`int()`, `float()`, `str()`, `bool()` — функции преобразования типа. Они не меняют
исходное значение, а создают новое:


````python
n = 7
print(str(n) + " days")     # число -> строка, затем склейка
print(float(n))             # 7.0
print(int("2026"))          # строка -> число
print(int(3.99))            # дробная часть отбрасывается, а не округляется
````

## 5. Арифметика


````python
a = 17
b = 5

print(a + b)     # 22
print(a - b)     # 12
print(a * b)     # 85
print(a / b)     # 3.4   — обычное деление, результат ВСЕГДА float
print(a // b)    # 3     — целочисленное деление (отбрасывает остаток)
print(a % b)     # 2     — остаток от деления
print(a ** b)    # 1419857 — возведение в степень
````

Пара `//` и `%` — рабочая лошадка: ими раскладывают величину на разряды
(секунды в часы и минуты, копейки в рубли, индекс в строку и столбец). Мы будем
пользоваться ими постоянно.

Приоритет операций обычный: `**`, затем `*`, `/`, `//`, `%`, затем `+` и `-`.
Скобки работают как в математике — и лучше поставить лишние, чем потом искать ошибку.


````python
print(2 + 3 * 4)      # 14
print((2 + 3) * 4)    # 20
````

**Ловушка, о которой стоит знать сразу.** Числа с плавающей точкой хранятся в двоичном
виде и представляют десятичные дроби приближённо:


````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Это свойство стандарта IEEE 754, общее почти для всех языков.
Практический вывод: **не сравнивайте дробные числа на точное равенство**
и не храните деньги во `float` в настоящих финансовых системах (там используют
целые копейки или тип `Decimal`).


### Приём 7 в действии

Проверим на месте, как работает `//` с отрицательными числами. Сначала — **гипотеза**,
и только потом запуск.

> `17 // 5` даёт 3, `17 % 5` даёт 2. Значит, `-17 // 5` должно дать −3, а `-17 % 5` — −2.

Запишите свой вариант ответа и выполните ячейку.


````python
print(-17 // 5)
print(-17 % 5)
````

Гипотеза не подтвердилась — и это самый полезный исход эксперимента: он показывает,
что наша модель операции была неверной.

Python округляет результат `//` **вниз**, к меньшему числу, а не в сторону нуля,
а знак остатка совпадает со знаком делителя. Правило, которое связывает обе операции,
выполняется всегда:


````python
a, b = -17, 5
print((a // b) * b + a % b == a)
````

## 6. Строки и форматирование вывода

Строка — последовательность символов в кавычках, одинарных или двойных (разницы нет).


````python
text = "Think like a programmer"

print(len(text))            # длина в символах
print(text.upper())         # верхний регистр
print(text.split())         # разбиение по пробелам -> список слов
print(text[0], text[-1])    # первый и последний символ
````

`text.upper()` — вызов **метода**: функции, привязанной к объекту. Методы не меняют
строку (строки в Python неизменяемы), а возвращают новую.

Склеивать строки можно через `+`, но для вывода почти всегда удобнее **f-строки**:


````python
name = "world"
items = 4

print("Hello, " + name + "!")            # конкатенация
print(f"Hello, {name}! Items: {items}")  # f-строка: выражения в фигурных скобках
````

Внутри `{}` можно писать любое выражение, а после двоеточия — формат вывода:


````python
value = 17 / 5

print(f"{value}")         # как есть
print(f"{value:.2f}")     # два знака после запятой
print(f"{value:8.2f}")    # то же, выровнено по ширине 8 символов
print(f"{1234567:,}")     # разделители разрядов
````

## 8. Собираем программу

Возвращаемся к плану из раздела 2. Теперь у нас есть всё, чтобы заменить
комментарии кодом.


````python
# INPUT
unit_price = 250.0
quantity = 4
discount_rate = 0.10
tax_rate = 0.20

# PROCESS
subtotal = unit_price * quantity
discount = subtotal * discount_rate
taxable = subtotal - discount
tax = taxable * tax_rate
total = taxable + tax

# OUTPUT
print(f"Subtotal: {subtotal:10.2f}")
print(f"Discount: {discount:10.2f}")
print(f"Tax:      {tax:10.2f}")
print(f"Total:    {total:10.2f}")
````

Три блока — три роли: получить вход, преобразовать, отдать выход. Каждая строка
блока PROCESS соответствует одному пункту плана.

Заметьте, что происходит при изменении требований. «А если скидка зависит от количества?»
— меняется одна строка, третья. «А если налог не начисляется?» — одна строка, пятая.
Это и есть практическая ценность декомпозиции: изменение требования затрагивает
ограниченный участок кода.

**Попробуйте сами:** измените `quantity` на 12 и `discount_rate` на `0.25`,
перезапустите ячейку.

### Четыре ошибки, которые вы будете видеть чаще всего

| Тип | Что означает | Типичная причина |
|---|---|---|
| `NameError` | Имя не определено | Опечатка в имени; ячейка с определением не выполнена |
| `TypeError` | Операция не подходит для этого типа | Смешали строку и число |
| `ValueError` | Тип верный, а значение — нет | `int("twelve")` |
| `SyntaxError` | Код не разбирается как Python | Незакрытая скобка или кавычка, пропущенное двоеточие |

`SyntaxError` стоит особняком: остальные возникают **во время** исполнения, а синтаксическая
ошибка обнаруживается до запуска — поэтому ячейка не выполнится вообще, включая корректные
строки в ней. И указывает интерпретатор на то место, где он заметил неладное, а это
нередко строка **после** настоящей ошибки: незакрытая скобка «съедает» следующую строку.

Проверьте это сами: снимите комментарий с первой строки в ячейке ниже и запустите.
